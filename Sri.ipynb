```python
print("======================================")
print("       PORTFOLIO RISK CLASSIFIER")
print("======================================")

stocks = float(input("Enter Stocks percentage: "))
mutual_funds = float(input("Enter Mutual Funds percentage: "))
bonds = float(input("Enter Bonds percentage: "))
average_return = float(input("Enter Average Return percentage: "))
volatility = float(input("Enter Volatility percentage: "))
```

```python
total_percentage = stocks + mutual_funds + bonds

print("\nTotal Portfolio Percentage:", total_percentage, "%")

if abs(total_percentage - 100) < 0.01:
    print("✅ Portfolio percentage is correct.")
else:
    print("❌ Error: Stocks + Mutual Funds + Bonds must equal 100%.")
```

```python
# Risk score calculation

risk_score = (
    (volatility * 0.55)
    - (average_return * 0.25)
    + (stocks * 0.20)
)

print("\nRisk Score:", round(risk_score, 2))
```
